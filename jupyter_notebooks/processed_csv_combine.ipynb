# Combine processed CSV files

Combine every CSV in `data/processed` and write the result to `data/processed/combined/combined.csv`.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "data/processed").is_dir()
)
INPUT_DIR = PROJECT_ROOT / "data/processed"
OUTPUT_DIR = INPUT_DIR / "combined"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

csv_files = sorted(INPUT_DIR.glob("*.csv"))
combined = pd.concat((pd.read_csv(path) for path in csv_files), ignore_index=True)
output_path = OUTPUT_DIR / "combined.csv"
combined.to_csv(output_path, index=False)

print(f"Combined {len(csv_files)} files into {output_path}")
```
